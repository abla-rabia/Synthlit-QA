```python
import json
from typing import List

#Unstructured for document parsing
from unstructured.partition.pdf import partition_pdf

```

```python
def partitionning_documents(doc_path=""):
    elements=partition_pdf(
        filename=doc_path, 
        strategy="hi_res",
        infer_table_structure=True
    )
    return(elements)

doc_path=r"..\data\pdf\2022_zimmermann_mri-sequence-independent-head-sct-proton.pdf"
elements=partitionning_documents(doc_path)
elements
```

```python
for i, el in enumerate(elements):
    print(i, el.category, el.metadata.page_number, el.text[:80])
```
